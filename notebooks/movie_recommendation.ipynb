Genre Extraction

```python
import ast

def extract_names(text):
  items=ast.literal_eval(text)
  names=[]
  for item in items:
    names.append(item["names"])
    return names
```

```python
def extract_names(text):
  items =ast.literal_eval(text)
  names=[]

  for item in items:
     names.append(item["name"])

     return names
```

```python
def convert_to_string(value):

    # If value is a list, join its elements
    if isinstance(value, list):
        return " ".join(value)

    # If value is missing or not a list
    return ""
```

```python
print(movies["overview"].iloc[0])
```

```python

movies["genres"] = movies["genres"].apply(
    lambda x: " ".join(x)
)
movies["keywords"] = movies["keywords"].apply(
    lambda x: " ".join(x)
)

movies["tags"] = (
    movies["genres"] + " " +
    movies["keywords"] + " " +
    movies["overview"]
)

print(movies[["title", "tags"]].head())
```
